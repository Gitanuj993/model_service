Feature Engineering

```python
cost_features = [
    "ministry",
    "sector",
    "agency",
    "state",
    "original_cost_cr",
    "cumulative_expenditure_cr",
    "physical_progress_pct",
    "financial_progress_pct",
    "progress_gap",
    "start_delay_months"
]

target = "cost_overrun"
```

Add them into Data Frame

```python
financial_progress_pct = (
    cumulative_expenditure_cr / original_cost_cr
) * 100

progress_gap = (
    financial_progress_pct - physical_progress_pct
)

start_delay_months = (
    revised_start_date - approval_start_date
).days / 30.44
```

Run This

```python

# 1. Financial progress
df["financial_progress_pct"] = (
    df["cumulative_expenditure_cr"]
    / df["original_cost_cr"].replace(0, np.nan)
) * 100
```

```python
# 2. Progress gap
df["progress_gap"] = (
    df["financial_progress_pct"]
    - df["physical_progress_pct"]
)
```

Clean the date time

```python
df["approval_start_date"] = pd.to_datetime(
    df["approval_start_date"], format="%m/%Y", errors="coerce"
)

df["revised_start_date"] = pd.to_datetime(
    df["revised_start_date"], format="%m/%Y", errors="coerce"
)
```

```python
# 3. Start delay in months
df["start_delay_months"] = (
    (df["revised_start_date"].dt.year -
     df["approval_start_date"].dt.year) * 12
    +
    (df["revised_start_date"].dt.month -
     df["approval_start_date"].dt.month)
)
```

# Convert dates

```python
df["approval_start_date"] = pd.to_datetime(
    df["approval_start_date"],
    format="%m/%Y",
    errors="coerce"
)
```

```python
df["revised_start_date"] = pd.to_datetime(
    df["revised_start_date"],
    format="%m/%Y",
    errors="coerce"
)
```

Convert String to Numeric values

```python
# Make physical progress numeric
df["physical_progress_pct"] = pd.to_numeric(
    df["physical_progress_pct"],
    errors="coerce"
)
```

# Prepare X and Y

```python
X = df[cost_features]
y = df["cost_overrun"]
```

prepare  train/test split

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)
```

 Define Numerical and categorical columns

```python
categorical_features = [
    "ministry",
    "sector",
    "agency",
    "state"
]

numeric_features = [
    "original_cost_cr",
    "cumulative_expenditure_cr",
    "physical_progress_pct",
    "financial_progress_pct",
    "progress_gap",
    "start_delay_months"
]
```

# Preprocessing pipeline

Import requirements

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
```

Now Convert

```python
numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])
```

Create Preprocessor

```python
preprocessor = ColumnTransformer([
    ("numeric", numeric_pipeline, numeric_features),
    ("categorical", categorical_pipeline, categorical_features)
])
print("done")
```

Create Model : Logistic Regression

```python
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline

cost_model = Pipeline([
    ("preprocessing", preprocessor),
    ("model", LogisticRegression(
        class_weight="balanced",
        max_iter=1000
    ))
])
```

# Train Model

```python
cost_model.fit(X_train, y_train)

print("Cost Overrun Model trained successfully.")
```

Prediction Checking

```python
y_pred = cost_model.predict(X_test)

y_prob = cost_model.predict_proba(X_test)[:, 1]

print(y_pred[:10])
print(y_prob[:10])
```

Model evaluation

```python
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score

print("Classification Report:")
print(classification_report(y_test, y_pred))

print("ROC-AUC:")
print(roc_auc_score(y_test, y_prob))

print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```

# Model Ready to Save

```python
import joblib

joblib.dump(cost_model, "cost_overrun_model_v1.pkl")

print("Model saved successfully.")
```
